```python
import shutil
import os

def vaciar_carpeta(ruta_carpeta):
    # Verificamos si la ruta existe
    if os.path.exists(ruta_carpeta):
        for elemento in os.listdir(ruta_carpeta):
            ruta_completa = os.path.join(ruta_carpeta, elemento)
            try:
                if os.path.isfile(ruta_completa) or os.path.islink(ruta_completa):
                    os.unlink(ruta_completa)  # Elimina archivos o enlaces simbólicos
                elif os.path.isdir(ruta_completa):
                    shutil.rmtree(ruta_completa) # Elimina subcarpetas y su contenido
                print(f"Eliminado: {elemento}")
            except Exception as e:
                print(f"Error al eliminar {elemento}: {e}")
    else:
        print("La ruta especificada no existe.")

# Uso:
vaciar_carpeta('/content/GSE144153_FAIR_RNAseq/figures')
vaciar_carpeta('/content/GSE144153_FAIR_RNAseq/processed')
vaciar_carpeta('/content/GSE144153_FAIR_RNAseq/metadata')
#vaciar_carpeta('/content/data')
```
